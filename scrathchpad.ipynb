```python
import polars as pl
from pathlib import Path

# Quick exploration: just read a few anchor files directly
triplets_dir = Path("./data/naics_training_pairs")
sample_files = list(triplets_dir.glob("*/*.parquet"))[:5]  # Just 5 anchors

df = pl.read_parquet(sample_files)
print(f"Loaded {len(df):,} rows from {len(sample_files)} files")
df
```

```python
# Quick stats on the sample
print(f"Columns: {df.columns}")
print(f"\nUnique anchors: {df['anchor_idx'].n_unique()}")
print(f"Unique positives: {df['positive_idx'].n_unique()}")
print(f"Unique negatives: {df['negative_idx'].n_unique()}")
print(f"\nMargin stats:")
df.select("relation_margin", "distance_margin").describe()
```
